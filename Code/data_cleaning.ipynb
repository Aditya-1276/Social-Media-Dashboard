#### Instagram Data

```python
insta_data = pd.read_excel(common_path / 'Data' / 'raw' / 'dataset_instagram-scraper_2026-05-13_14-35-13-924.xlsx')
```

```python
insta_data.describe()
```

```python
insta_data.shape
```

```python
list(insta_data.columns)
```

```python
# dumping all of the unnecessary columns apify gave us. 
# Most of these are just useless for our analysis and 
# some of them have too many nulls and too few unique values to be useful.
keep = [
    'shortCode', 'url', 'type', 'timestamp',
    'likesCount', 'commentsCount',
    'caption',
] + [f'hashtags/{i}' for i in range(11)]

keep = [c for c in keep if c in insta_data.columns]
insta_data = insta_data[keep]
print(insta_data.shape)
```

```python
insta_data.head()
```

```python
print(insta_data.isnull().sum())
```

```python
# collapse all hashtag values into a single column
hashtag_cols = [c for c in insta_data.columns if c.startswith('hashtags/')]
insta_data['hashtags'] = insta_data[hashtag_cols].apply(lambda row: ', '.join(row.dropna().astype(str)), axis=1)
insta_data.drop(columns=hashtag_cols, inplace=True)
```

```python
print(insta_data.shape)
print(insta_data.isnull().sum())
```

```python
insta_data.head(5)
```

```python
insta_data.info()
```

```python
print(insta_data.describe())
```

```python
mean, std = insta_data['likesCount'].mean(), insta_data['likesCount'].std()
print(f"INSTAGRAM\nLikes\nMean: {mean}, Std: {std}")
mean, std = insta_data['commentsCount'].mean(), insta_data['commentsCount'].std()
print(f"Comments\nMean: {mean}, Std: {std}")
```

```python
insta_data.to_excel(common_path / 'Data' / 'cleaned' / 'instagram_clean.xlsx', index=False)
print("Done:", insta_data.shape)
```

#### X Data

```python
tweet_data = pd.read_excel(common_path / 'Data' / 'raw' / 'dataset_twitter-x-data-tweet-scraper-pay-per-result-cheapest_2026-05-13_15-20-30-100.xlsx')
```

```python
tweet_data.shape
```

```python
keep = [
    'url', 'twitterUrl', 'createdAt',
    'text', 'lang',
    'likeCount', 'replyCount', 'retweetCount', 
    'quoteCount', 'viewCount', 'bookmarkCount',
    'isReply', 'isPinned',
    'entities/hashtags/0/text',
    'entities/hashtags/1/text',
    'entities/hashtags/2/text',
]

keep = [c for c in keep if c in tweet_data.columns]
tweet_data = tweet_data[keep]
print(tweet_data.shape)
print(tweet_data.isnull().sum())
```

```python
htag_cols = [c for c in tweet_data.columns if c.startswith('entities/hashtags')]
tweet_data['hashtags'] = tweet_data[htag_cols].apply(lambda r: ', '.join(r.dropna().astype(str)), axis=1)
tweet_data.drop(columns=htag_cols, inplace=True)
```

```python
tweet_data.head()
```

```python
tweet_data[tweet_data['isReply'] == True].sum()
```

```python
# Since none of the tweets are replies, we can drop this column.
tweet_data.drop(columns=['isReply'], inplace=True)
```

```python
tweet_data.isnull().sum()
```

```python
tweet_data.duplicated().sum()
```

```python
tweet_data.drop_duplicates(inplace=True)
tweet_data.shape
```

```python
tweet_data.describe()
```

```python
tweet_data.info()
```

```python
mean, std = tweet_data['likeCount'].mean(), tweet_data['likeCount'].std()
print(f"TWITTER\nLikes\nMean: {mean}, Std: {std}")
mean, std = tweet_data['replyCount'].mean(), tweet_data['replyCount'].std()
print(f"Replies\nMean: {mean}, Std: {std}")
mean, std = tweet_data['retweetCount'].mean(), tweet_data['retweetCount'].std()
print(f"Retweets\nMean: {mean}, Std: {std}")
mean, std = tweet_data['quoteCount'].mean(), tweet_data['quoteCount'].std()
print(f"Quotes\nMean: {mean}, Std: {std}")
mean, std = tweet_data['viewCount'].mean(), tweet_data['viewCount'].std()
print(f"Views\nMean: {mean}, Std: {std}")
mean, std = tweet_data['bookmarkCount'].mean(), tweet_data['bookmarkCount'].std()
print(f"Bookmarks\nMean: {mean}, Std: {std}")
```

```python
tweet_data.to_excel(common_path / 'Data' / 'cleaned' / 'twitter_clean.xlsx', index=False)
print("Done:", tweet_data.shape)
```
